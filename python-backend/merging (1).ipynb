```python
persona_df = pd.read_csv('student_personas.csv')
display(persona_df.head(2))
grade_df = pd.read_csv('grade_final.csv')
display(grade_df.head(2))
attendance_df = pd.read_csv('monthly_subject_attendance.csv')
display(attendance_df.head(2))
back_df = pd.read_csv('student_backlogs_v2.csv')
display(back_df.head(2))
```

#agregate grades data

```python
# Define the columns containing marks
marks_cols = ['Quiz_1', 'Quiz_2', 'Quiz_3', 'Midterm']

# Calculate the average score for each course
grade_df['Course_Performance_Score'] = grade_df[marks_cols].mean(axis=1)

# Now, group by student and calculate their average score across all courses
student_grades_agg = grade_df.groupby('studentID')['Course_Performance_Score'].mean().reset_index()
student_grades_agg = student_grades_agg.rename(columns={'Course_Performance_Score': 'Semester_Performance_Score'})

print("--- Aggregated Student Grades ---")
print(student_grades_agg)
```

## Aggregate Attendance data


```python
import pandas as pd

# 1. DEFINE the DataFrame first
# This creates the sample attendance data.
# If you have an Excel file, you would use: attendance_df = pd.read_excel('your_file.xlsx')
attendance_data = {
    'studentID': ['E0001', 'E0001', 'E0001', 'E0002', 'E0002'],
    'CourseID': ['CSET243', 'CSET243', 'CSET240', 'CSET243', 'CSET250'],
    'Month': ['July', 'August', 'July', 'July', 'August'],
    'Attendance_Percentage': [95, 98, 92, 65, 70]
}
attendance_df = pd.DataFrame(attendance_data)


# 2. NOW, run your aggregation code
# This part of your code is correct, it just needed the DataFrame to exist.
student_attendance_agg = attendance_df.groupby('studentID')['Attendance_Percentage'].mean().reset_index()
student_attendance_agg = student_attendance_agg.rename(columns={'Attendance_Percentage': 'Avg_Attendance_Percentage'})


# 3. Print the result
print("--- Aggregated Student Attendance ---")
print(student_attendance_agg)
```

# Aggregate Backlogs data

```python
import pandas as pd

# 1. Define the backlog_df DataFrame
backlog_data = {
    'studentID': ['E0001', 'E0001', 'E0002', 'E0002'],
    'Course_code': ['CSET243', 'CSET240', 'CSET243', 'CSET250'],
    'Backlog': [0, 0, 1, 0],
    'persona': ['Fast learners', 'Fast learners', 'Slow learners', 'Slow learners']
}
backlog_df = pd.DataFrame(backlog_data)

# 2. Now, you can run the aggregation code on the defined DataFrame
student_backlog_agg = backlog_df.groupby('studentID')['Backlog'].sum().reset_index()
student_backlog_agg = student_backlog_agg.rename(columns={'Backlog': 'Total_Backlogs'})

print("\n--- Aggregated Student Backlogs ---")
print(student_backlog_agg)
```

# **Merging the two semesters**

```python
import pandas as pd

# This is the DataFrame you already have from the current semester
current_semester_df = pd.DataFrame({
    'studentID': ['E0001', 'E0002', 'E0003'],
    'Current_CGPA': [9.07, 6.50, 7.20],
    'Avg_Attendance_Percentage': [95.0, 67.5, 75.0],
    'Total_Backlogs': [0, 1, 3]
})

# This is the data you generate from the PREVIOUS semester's files
previous_semester_df = pd.DataFrame({
    'studentID': ['E0001', 'E0002', 'E0003'],
    'Current_CGPA': [8.95, 6.95, 7.30],
    'Avg_Attendance_Percentage': [96.0, 85.0, 76.0],
    'Total_Backlogs': [0, 0, 3]
})
```

```python
# Merge the two dataframes on the studentID
# The suffixes distinguish between the current and previous semester's data
trend_df = pd.merge(
    current_semester_df,
    previous_semester_df,
    on='studentID',
    suffixes=('_current', '_previous')
)

print("--- Merged Semester Data ---")
print(trend_df)
```

# TREND **FEATURES**

```python
# Calculate the difference between current and previous semester metrics
trend_df['CGPA_Trend'] = trend_df['Current_CGPA_current'] - trend_df['Current_CGPA_previous']
trend_df['Attendance_Trend'] = trend_df['Avg_Attendance_Percentage_current'] - trend_df['Avg_Attendance_Percentage_previous']
trend_df['Backlog_Trend'] = trend_df['Total_Backlogs_current'] - trend_df['Total_Backlogs_previous']

print("\n--- Data with Trend Features ---")
print(trend_df)
```

**Final Model Ready Dataset**

```python
# Select the most important columns for your final dataset
final_model_df = trend_df[[
    'studentID',
    'Current_CGPA_current',
    'Avg_Attendance_Percentage_current',
    'Total_Backlogs_current',
    'CGPA_Trend',
    'Attendance_Trend',
    'Backlog_Trend'
]]

# Rename current columns for clarity
final_model_df = final_model_df.rename(columns={
    'Current_CGPA_current': 'Current_CGPA',
    'Avg_Attendance_Percentage_current': 'Current_Attendance',
    'Total_Backlogs_current': 'Current_Backlogs'
})


print("\n--- Final Model-Ready Dataset ---")
print(final_model_df)
```
